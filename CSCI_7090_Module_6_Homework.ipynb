## Q1 — Build & train a small CNN (15 pts)
Build a small CNN for Fashion-MNIST (input `28×28×1`, 10 output classes) with at least one `Conv2D`, a pooling layer, and dense layers; compile and train it (a few epochs is fine), then report **test accuracy**.

**Produce:** the fitted `model` and `acc` (test accuracy)

```python
# Build, compile, and train a CNN on (xtr, ytr); evaluate on (xte, yte).
# Produce: model, acc

model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(16, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(32, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(xtr, ytr, epochs=10, batch_size=64, validation_split=0.1, verbose=1)

loss, acc = model.evaluate(xte, yte, verbose=0)
print("Test Accuracy:", acc)
```

## Q7 — Short answer, no code (20 pts)

**(a)** Why is a fully-connected layer applied **directly** to a `128×128×3` image impractical? Name the single key observation about images that convolution exploits to fix this.

> **Answer:** Applying a fully-connected layer directly is impractical because of the massive number of weights required. An input of $128 \times 128 \times 3 = 49,152$ pixels connected to a moderate hidden layer of, say, 1,000 neurons results in nearly 50 million parameters for just a single layer, leading to severe overfitting and extremely high memory/computational cost.
>
> The single key observation about images that convolution exploits is **spatial locality** (the idea that nearby pixels are highly correlated and form local features, allowing us to share weights locally rather than connecting every pixel globally).

**(b)** Why is **ReLU** applied after convolutional layers? What would go wrong if a deep CNN used **no** activation function at all between its conv layers?

> **Answer:** **ReLU** is applied to introduce non-linearity into the network. Without non-linear activation functions, the composition of multiple linear operations (convolutions are linear transformations) remains a simple linear function. Therefore, a deep CNN with no activation functions would behave exactly like a shallow, single-layer linear model, rendering depth useless and making it unable to learn complex patterns.

**(c)** Explain the difference between **'valid'** and **'same'** padding, and give **one advantage and one disadvantage** of each.

> **Answer:**
> - **'valid' padding** means no padding is added. The filter only slides over valid pixels of the input.
>   - *Advantage:* Highly computationally efficient and prevents artifact creation at the borders.
>   - *Disadvantage:* The spatial dimensions shrink with each layer, and edge pixels are under-represented because they are rarely processed by the filter.
> - **'same' padding** adds zero-padding symmetrically around the boundaries so that the output spatial shape matches the input spatial shape (when stride is 1).
>   - *Advantage:* Preserves spatial dimensions, allowing for very deep architectures without premature shrinkage of feature maps.
>   - *Disadvantage:* Introduces zero-padding boundary artifacts (since zeros are padded around the real data, which can distort features near the edges).
