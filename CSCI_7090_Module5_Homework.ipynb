## Setup A — run first
Loads the **breast-cancer** dataset into `X, y` and generates your personal parameters (`TEST_SIZE`, `Z_VAL`, `COEF_IDX`).

```python
# === SETUP A — run first ===
ROLL_NUMBER = "901503252"
import hashlib, numpy as np
from sklearn.datasets import load_breast_cancer, load_digits
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

seed = int(hashlib.sha256(ROLL_NUMBER.encode()).hexdigest(), 16) % (2**32)
rng  = np.random.default_rng(seed)

# Breast-cancer dataset -> X, y
X, y = load_breast_cancer(return_X_y=True)

# Your personal parameters (unique to your roll number) — DO NOT EDIT
TEST_SIZE = float(rng.choice([0.2, 0.25, 0.3]))
Z_VAL     = round(float(rng.uniform(-4, 4)), 2)
COEF_IDX  = int(rng.integers(0, 30))

print("Roll:", ROLL_NUMBER, "| seed:", seed)
print(f"TEST_SIZE={TEST_SIZE}  Z_VAL={Z_VAL}  COEF_IDX={COEF_IDX}")
```

## Q1 — Train a binomial logistic regression (15 pts)
Using your `TEST_SIZE`, split `X, y` into train/test (`random_state = seed`, `stratify = y`), then train a `LogisticRegression` (`max_iter = 10000`).

**Produce:** `Xtr, Xte, ytr, yte` and the fitted model `clf`

```python
# Split X, y with TEST_SIZE (random_state=seed, stratify=y); then train LogisticRegression(max_iter=10000).
# Produce: Xtr, Xte, ytr, yte, clf

Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=TEST_SIZE, random_state=seed, stratify=y)
clf = LogisticRegression(max_iter=10000).fit(Xtr, ytr)
```

## Q2 — Report classification metrics (15 pts)
For `clf` on the test set, compute and print **accuracy, precision, recall,** and **F1** (write the metric calls yourself, e.g. from `sklearn.metrics`).

**Produce:** `acc, prec, rec, f1`

```python
# Predict on Xte, then compute the four metrics for clf.
# Produce: acc, prec, rec, f1

y_pred = clf.predict(Xte)
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

acc = accuracy_score(yte, y_pred)
prec = precision_score(yte, y_pred)
rec = recall_score(yte, y_pred)
f1 = f1_score(yte, y_pred)

print("Accuracy:", acc)
print("Precision:", prec)
print("Recall:", rec)
print("F1:", f1)
```

## Q3 — Sigmoid (10 pts)
Write a sigmoid function `σ(z) = 1 / (1 + e^(−z))` and evaluate it at your `Z_VAL`.

**Produce:** `sig` (print it)

```python
# Define sigmoid(z), then evaluate it at Z_VAL.
# Produce: sig

def sig(z):
    return 1 / (1 + np.exp(-z))

sig = sig(Z_VAL)
print("σ(Z_VAL):", sig)
```

**Written:** Which class (0 or 1) does a 0.5 threshold assign to your result, and why?

Because $\sigma(Z\_VAL) = \sigma(0.29) \approx 0.572$, which is greater than the 0.5 threshold ($0.572 > 0.5$), this result is assigned to **class 1**.

```python
# Digits dataset -> X2, y2
X2, y2 = load_digits(return_X_y=True)
print("X2:", X2.shape, "| classes:", len(np.unique(y2)))
```

## Q4 — Train a multinomial logistic regression (15 pts)
Split `X2, y2` with your `TEST_SIZE` (`random_state = seed`, `stratify = y2`) and train a `LogisticRegression` (`max_iter = 10000`).

**Produce:** `X2tr, X2te, y2tr, y2te` and the fitted model `mclf`

```python
# Split X2, y2 with TEST_SIZE (random_state=seed, stratify=y2); then train LogisticRegression(max_iter=10000).
# Produce: X2tr, X2te, y2tr, y2te, mclf

X2tr, X2te, y2tr, y2te = train_test_split(X2, y2, test_size=TEST_SIZE, random_state=seed, stratify=y2)
mclf = LogisticRegression(max_iter=10000).fit(X2tr, y2tr)
```

## Q5 — Report metrics, macro-averaged (15 pts)
For `mclf` on the digits test set, compute **accuracy**, and **macro-averaged** precision, recall, and F1 (use `average='macro'`).

**Produce:** `macc, mprec, mrec, mf1`

```python
# Predict on X2te, then compute accuracy and the three macro-averaged metrics for mclf.
# Produce: macc, mprec, mrec, mf1

y2_pred = mclf.predict(X2te)
macc = accuracy_score(y2te, y2_pred)
mprec = precision_score(y2te, y2_pred, average="macro")
mrec = recall_score(y2te, y2_pred, average="macro")
mf1 = f1_score(y2te, y2_pred, average="macro")

print("Accuracy:", macc)
print("Precision:", mprec)
print("Recall:", mrec)
print("F1:", mf1)
```

## Q6 — Coefficients & probability (15 pts)
From `clf` (your binomial model), read coefficient number `COEF_IDX`, and predict the probability of class 1 for test sample 0 (`Xte[0]`).

**Produce:** `coef_val, prob0`

```python
# Read clf's coefficient at index COEF_IDX, and the class-1 probability for Xte[0].
# Produce: coef_val, prob0
coef_val = clf.coef_[0][COEF_IDX]
prob0 = clf.predict_proba(Xte[0].reshape(1, -1))[0][1]

print("Coefficient value:", coef_val)
print("Class-1 probability:", prob0)
```

## Q7 — Short answer, no code (15 pts)
Answer in your own words:

**(a)** Logistic regression is often called a "single-neuron neural network." In that picture, what do the *coefficients*, the *intercept*, and the *sigmoid* each correspond to?

* **Coefficients (weights):** They correspond to the connection weights of the incoming synapses, representing the strength and direction of each input feature's influence.
* **Intercept (bias):** It corresponds to the bias term of the neuron, which shifts the activation threshold.
* **Sigmoid:** It corresponds to the non-linear activation function of the neuron, squashing the raw weighted sum into an output between 0 and 1.

**(b)** The model is trained by minimizing **log-loss**, not by maximizing accuracy directly. Give one concrete reason why.

Accuracy is a step function (discontinuous and flat almost everywhere) because a small change in weights may not change the thresholded predicted class, making its gradient zero. Log-loss, on the other hand, is a smooth, continuous, and convex function with informative gradients, which allows gradient-based optimization algorithms to converge reliably.

**(c)** Your binomial model outputs a probability. Describe how it turns one image's features into a final 0/1 label — name the two steps (linear score → probability) and where the **0.5 threshold** fits in.

1. **Linear Score Calculation:** The model computes a raw real-valued score (logit) using the linear combination of features, weights, and bias: $z = w^T x + b$.
2. **Probability Mapping (Sigmoid):** This score $z$ is passed through the sigmoid function to map it to a probability $p = \sigma(z) \in (0, 1)$.
3. **Decision Threshold:** The final 0/1 label is determined by comparing $p$ to the **0.5 threshold**: if $p \ge 0.5$, the label is 1; otherwise, it is 0.
